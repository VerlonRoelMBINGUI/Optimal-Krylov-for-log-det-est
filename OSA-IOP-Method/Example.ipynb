```python
import scipy.io
import scipy.sparse as sp
import time
import warnings
warnings.filterwarnings("ignore")
from intervals import *
from arnoldi_iop_log import *
from hutchpp import *
from osa_iop import *


def load_matrix(path="ship_001.mat"):
    mat = scipy.io.loadmat(path)
    Q = mat["Problem"]["A"][0, 0]

    if not sp.isspmatrix_csr(Q):
        Q = sp.csr_matrix(Q)

    return Q


def main():
    Q = load_matrix()

    estimator = LogDetEstimator(
        Q,
        m=21,
        krylov_dim=30,
        tol=1e-6,
    )

    start = time.perf_counter()

    logdet, info = estimator.estimate()

    elapsed = time.perf_counter() - start

    print("\n========================")
    print(" LogDet Hutch++ Result")
    print("========================")
    print(f"log(det(Q)) ≈ {logdet:.6f}")
    print(f"Time (s)     = {elapsed:.4f}")
    print("info:", info)


if __name__ == "__main__":
    main()
```
