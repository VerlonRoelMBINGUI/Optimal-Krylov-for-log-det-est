```python

import time
import numpy as np

import scipy.io
from utils import *
from operators import *
from lanczos import *
from Oslq import *

def load_suite_sparse_mat(path):
    mat_data = scipy.io.loadmat(path)
    problem = mat_data["Problem"]
    Q = problem["A"][0, 0]
    return Q

# Load matrix
Q = load_suite_sparse_mat("crystm02.mat")

estimator = SLQLogDetEstimator(Q)

rng = np.random.default_rng(42)

start = time.time()

ld, breakdown = estimator.estimate(
    m=20,
    n_residual_vecs=10,
    s=30,
    q_power=1,
    rng=rng,
    return_breakdown=True,
)

print("logdet estimate:", ld)
print("alpha:", breakdown["alpha"])
print("subspace:", breakdown["subspace"])
print("residual:", breakdown["residual"])
print("time:", time.time() - start)
```
